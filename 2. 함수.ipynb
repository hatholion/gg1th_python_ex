## 함수 문제 1

```python
def mul (*values):
    #print(values)
    #print(type(values)) # *values는 tuple 형태로 받는다.
    val = 1
    for i in values:
        val = val * i
    return val

print(mul(5, 7, 9, 10))
```

```python
mul(5,7,9,10,11)
```

```python
result = mul(5,7,9,10,11,13)
print(result)
```

```python
# 해당 코드는 소수 두 개들 넣었을 때에만 유효함. 최소공배수 mul을 num1, num2의 곱으로 가정하므로.
def com_mul(num1, num2):
    mul = num1 * num2
    for i in range(1, 101):
        if i % mul == 0:
            print(f"{i}은(는) {num1}과 {num2}의 공배수입니다.")

a, b = input('두 수 입력:').split(",")
a = int(a)
b = int(b)
#print(a, b)
com_mul(a, b)
```

```python
# 최소공배수 구해서 하는 법
def comMul(a, b):
    for i in range(1, 101):
        if i % a == 0 and i % b == 0:
            print(f"{i}은(는) {a}와 {b}의 공배수입니다.")

a, b = map(int, input("두 수 입력 예) 2, 3: ").split(","))

comMul(a, b)
```

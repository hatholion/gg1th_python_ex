## 반복문 - for문

```python
numbers = [273, 103, 5, 32, 65, 9, 72, 800, 99, 19]
len(numbers)
```

```python
# 100 이상의 숫자만 출력
for num in numbers:
    #반복
    # 100 이상인 것만 조건문으로 참으로 설정해 출력
    if num >= 100:
        print(f" - 100 이상의 수 : {num}")
```

```python
numbers
```

```python
# 100 이상의 숫자, 100 미만인 숫자를 
# 새로운 리스트에 넣어 각각 리스트로 출력
a = []
b = []
for num in numbers:
    if num >= 100:
        a.append(num)
    else:
        b.append(num)
print(f" - 100 이상의 수: {a}")
print(f" - 100 미만의 수: {b}")
```

```python
# 실습자 정답
# 100 이상의 숫자, 100 미만인 숫자를 새로운 리스트에 넣어 
#각각 리스트로 출력
# 리스트 초기화는 반복문 안에 들어갈 경우 하나 들어간 뒤 초기화됨.
lower100 = []
higher100 = [] 
for num in numbers:
    if num >= 100:
        #100 이상 리스트 추가
        higher100.append(num)
    else:
        # 100 미만 리스트 추가
        lower100.append(num)

print(f" - 100 이상의 수: {higher100}")
print(f" - 100 미만의 수: {lower100}")
```

## for + dict

```python
# nums의 숫자가 몇 번 등장하는지 dict 구조로 출력하는 코드
#{1:횟수, 2: ,3: ,4: ,5: , ...}
nums = [1, 2, 6, 7, 4, 3, 2, 1, 5, 9, 4, 9, 7, 2, 1, 3, 4, 5, 6, 7, 5, 6, 7, 5]
```

```python
a = {}
a[0] = 1 # dictionary의 create
a[1] = 1
a[2] = 2
a[3] = 3
a[1] = 2*3
a
```

```python
counter = {}

for num in nums: 
    # key 값 판단 : counter dictionary 안에 이미 있는지로 판단
    # 그 숫자가 제일 처음 나왔을 때, key 값 + 카운트(초기화)
    if num not in counter:
        counter[num] = 1
    # 그 숫자가 처음 나온 것이 아닐 때, 카운트
    else:
        counter[num] += 1

counter
```

## List comprehension

```python
[i * i for i in range(0,20,2)]

# 예측 결과: [0, 4, 16, ...]
```

```python
array = [i * i for i in range(0,20,2)]
array
```

```python
# for문으로 만들기
array = []
for i in range(0, 20, 2):
    array.append(i*i)
print(array) # = array
```

### List Comprehension의 조건문

```python
even_nums = []
for i in range(1,21):
    # i가 짝수인지 판단, 짝수이면 리스트에 추가
    if i%2 == 0:
        even_nums.append(i)
even_nums
```

```python
# 짝수만 출력하기
array = [i for i in range(1,21) if i%2 == 0]
array
```

## 제어문 종합 문제

```python
decimal = []
for num in range (1, 101):
    if num == 2:
        decimal.append(num)
    deno = num - 1
    
    for i in range(2, deno+1):
        if num % i == 0:
            break
        elif i == deno:
            decimal.append(num)

print("1~100까지의 소수 결과\n", decimal)
```

```python
decimal = []
for num in range (2, 101):
    if num == 2:
        decimal.append(num)
    deno = num - 1
    
    while ((deno/2) >= 1): 
        if num % deno == 0:
            break
        else:
            deno = deno - 1

        if deno == 1:
            decimal.append(num)

print("1~100까지의 소수 결과\n", decimal)
```

```python
decimal = []
for num in range(2, 101):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        decimal.append(num)

print("1~100까지의 소수 결과\n", decimal)
```
